```python
def dropnas(df): 
    nas = df.isna()
    num = df.isna().sum()
    percent = (df.isna().mean() * 100)
    missing_df = pd.DataFrame({'Missing Count' : num, 'Missing Percentage': percent})
    
    return missing_df
```

```python
dropnas(df)
```

```python
def findnas(df):
    null = df.isna()
    count = df.isna().sum()
    percentage = (df.isna().mean() * 100)
    frame = pd.DataFrame({"Missing Count": count, "Missing Percentage": percentage})
    
    return frame
```

```python
test_df = findnas(df)
```
